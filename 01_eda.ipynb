
## Exploratory Data Analysis (EDA)
This notebook explores the healthcare cost dataset to understand the distribution of costs, key features, and potential predictors of high-cost individuals.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
df = pd.read_csv("insurance.csv")
df.head()
df.info()
df.describe()
df.isnull().sum()#缺失值检查
```

### Dataset Overview

The dataset contains individual-level demographic and health-related attributes, along with corresponding medical insurance charges.
No missing values are observed in the dataset.



##哪些变量是数值型
##有没有缺失值
##数据规模（样本量）

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["charges"], bins=50, kde=True)
plt.title("Distribution of Medical Insurance Charges")
plt.xlabel("Charges")
plt.ylabel("Count")
plt.show()
```

### Distribution of Healthcare Charges
The distribution of healthcare charges is highly right-skewed, indicating that a small proportion of individuals incur significantly higher medical costs.


### Smoking Status and Healthcare Charges


```python
plt.figure(figsize=(8, 5))
sns.boxplot(x="smoker", y="charges", data=df)
plt.title("Healthcare Charges by Smoking Status")
plt.show()
```

### BMI and Healthcare Charges


```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="bmi", y="charges", hue="smoker", data=df)
plt.title("BMI vs Charges by Smoking Status")
plt.show()
```

### EDA Summary

- Medical insurance charges exhibit a highly right-skewed distribution.
- Smoking status is strongly associated with higher healthcare costs.
- BMI shows a positive relationship with charges, particularly among smokers.
- These findings motivate further feature engineering and predictive modelling.
